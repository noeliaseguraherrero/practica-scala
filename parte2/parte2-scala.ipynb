## Ejercicio 6 — Decidir el ganador de una partida
- Si ambas manos superan 21, devuelve `0`: Utilice IF
- Si solamente `handA` supera 21, devuelve `handB`: Utilice ELSE IF
- Si solamente `handB` supera 21, devuelve `handA`.: Utilice ELSE IF
- Si ninguna supera 21, devuelve la mano con mayor puntuación: Utilice LESE y otro IF interno.

```scala
def ganador(handA: Int, handB: Int): Int = {
  if (handA > 21 && handB > 21) {
    0
  } else if (handA > 21 && handB < 21) {
    handB
  } else if (handB > 21 && handA < 21){
      handA
  } else {
      if(handA > handB){
          handA
      }else{
          handB
      }
  }
}
```

```scala
ganador(26,20)
ganador(18,22)
ganador(24,25)
ganador(17,19)
ganador(21,20)
```

## Ejercicio 12 — Operadores relacionales y lógicos

| Expresión | Resultado | Explicación |
| :--- | :--- | :--- |
| `handA > handB` | `false` | Comprueba si 18 es estrictamente mayor que 21. |
| `handB == 21` | `true` | Comprueba si 21 es  igual a 21. |
| `handC != 21` | `true` | Comprueba si 25 es diferente de 21. |
| `handA <= 21 && handB <= 21` | `true` | Evalúa con `&&` (Y) si ambas son menores o iguales a 21. |
| `handA > 21 \|\| handC > 21` | `true` | Utiliza `\|\|` (O) para comprobar si al menos una supera 21; como handC vale 25, la condición se cumple. |
| `!(handB > 21)` | `true` | Utilizamos el operador de negación `!` al resultado de evaluar si 21 es mayor que 21 (`!false` se convierte en `true`). |

```scala
val handA = 18
val handB = 21
val handC = 25
```

```scala
// 1. ¿Es handA mayor que handB?
val q1 = handA > handB
  
// 2. ¿Es handB igual a 21?
val q2 = handB == 21
  
// 3. ¿Es handC diferente de 21?
val q3 = handC != 21
  
// 4. ¿Son handA y handB menores o iguales a 21?
val q4 = handA <= 21 && handB <= 21
  
// 5. ¿Alguna de las manos handA o handC supera 21?
val q5 = handA > 21 || handC > 21
  
// 6. ¿No se ha pasado handB de 21?
val q6 = !(handB > 21)
```

## Ejercicio 13 — `foreach` y funciones como valores
- Cuál necesita un contador: La versión (While).
- Cuál necesita una variable `var` para recorrer la colección: La versión (While).
- Cuál se aproxima más al estilo funcional presentado en el material del curso (Datacamp): La versión B (foreach)

```scala
val manos = Array(17, 24, 21, 26, 18)
def bust(puntuacion: Int): Boolean = puntuacion > 21
```

```scala
println("Versión A")
var i = 0
  while (i < manos.length) {
    val mano = manos(i)
    println(s"Mano $mano -> ¿Es bust?: ${bust(mano)}")
    i += 1
  }

println("Versión B")
manos.foreach { mano =>
    println(s"Mano $mano -> ¿Es bust?: ${bust(mano)}")
  }
```

  if (seHaPasado) {

  }
}:* El bucle while se encarga de repetir el proceso mientras queden jugadores por revisar y los bloques if, else if y else evalúan las reglas: si el jugador se pasó, si sacó un 21 exacto, o si tiene otra puntuación válida, decidiendo qué mensaje mostrar y cuándo actualizar el récord.

- Qué parte de tu solución consideras más cercana al estilo funcional:

  *val seHaPasado = bust(mano):* Ya que no modifica variables de fuera.
- Qué parte consideras más cercana al estilo imperativo:

  *var i = 0 y while (i < jugadores.length) { ... if (mano > mejorPuntuacionValida) { mejorPuntuacionValida = mano} i += 1 }:* Vas modificando variables externas.

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)
def bust(puntuacion: Int): Boolean = puntuacion > 21
```

```scala
var mejorPuntuacionValida = -1
var i = 0
while (i < jugadores.length) {
    val jugador = jugadores(i)
    val mano = manos(i)

    val seHaPasado = bust(mano)

    if (seHaPasado) {
      println(s"- $jugador tiene $mano: Se ha pasado (Bust)")
    } else if (mano == 21) {
      println(s"- $jugador tiene $mano: ¡Perfecto (21)! Puntuación válida")
          if (mano > mejorPuntuacionValida) {
            mejorPuntuacionValida = mano
          }
    } else {
      println(s"- $jugador tiene $mano -> Puntuación válida")
      if (mano > mejorPuntuacionValida) {
        mejorPuntuacionValida = mano
      }
    }

    i += 1 
  }

if (mejorPuntuacionValida != -1) {
    println(s"\n La mejor puntuación válida es: $mejorPuntuacionValida")
  } else {
    println(s"\n No hay puntuaciones válidas.")
  }
```
